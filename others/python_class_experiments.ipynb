## Python Class 101

In this notebook we will experiment with Python classes. We will see use of `init` method, `super` function, class variables, class methods, static methods, inheritance, and more.


```python
## Lets create a basic class

class Animal:
    """_summary_
    """
    def __init__(self, name: str, animaltype:str) -> None:
        print("class is being init")
        self.name = name
        self.animaltype = animaltype

    def __str__(self) -> str:
        return(f"The animal is of name {self.name} and type {self.animaltype}")
        
```

```python
a1 = Animal(name="abc", animaltype="xyz")
print(a1)
```

```python
class Dog(Animal):
    """_summary_

    Args:
        Animal (_type_): _description_
    """
    def __init__(self, name: str, animaltype: str, sound:str) -> None:
        super().__init__(name, animaltype)
        self.sound = sound
    
    def __str__(self) -> str:
        return(f"The animal is of name {self.name} and type {self.animaltype} does sound {self.sound}")
```

```python
dog1 = Dog('freddy', 'dog', 'bark bark')
```

```python
print(dog1)
```
