
# Jacobian Convention and Shape Convention

## 1. **Introduction to the Jacobian**
The Jacobian matrix describes how a vector-valued function changes with respect to its inputs. For a function $\mathbf{f} : \mathbb{R}^n \to \mathbb{R}^m$, the Jacobian matrix is:

$$
\mathbf{J}_{\mathbf{f}} = \frac{\partial \mathbf{f}}{\partial \mathbf{x}} = 
\begin{bmatrix}
\frac{\partial f_1}{\partial x_1} & \cdots & \frac{\partial f_1}{\partial x_n} \\
\vdots & \ddots & \vdots \\
\frac{\partial f_m}{\partial x_1} & \cdots & \frac{\partial f_m}{\partial x_n}
\end{bmatrix},
$$

where $\mathbf{f} \in \mathbb{R}^m$, $\mathbf{x} \in \mathbb{R}^n$, and $\mathbf{J}_{\mathbf{f}} \in \mathbb{R}^{m \times n}$.

---

## 2. **Special Case: Vector Input to Scalar Output**

For a scalar-valued function $f : \mathbb{R}^n \to \mathbb{R}$, the Jacobian reduces to a **gradient row vector**:

$$
\mathbf{J}_f = \frac{\partial f}{\partial \mathbf{x}} = 
\begin{bmatrix}
\frac{\partial f}{\partial x_1} & \frac{\partial f}{\partial x_2} & \cdots & \frac{\partial f}{\partial x_n}
\end{bmatrix}.
$$

### Matrix Dimensions:
- $\mathbf{x} \in \mathbb{R}^n$,  
- $f \in \mathbb{R}$,  
- $\mathbf{J}_f \in \mathbb{R}^{1 \times n}$.

However, in machine learning, particularly during gradient descent, it is often more convenient to use the **shape convention**, where the gradient is reshaped to match the parameter's dimensions. For $ f : \mathbb{R}^n \to \mathbb{R} $, this involves representing the gradient as a **column vector**:

$$
\text{Shape Convention: } \nabla f = 
\begin{bmatrix}
\frac{\partial f}{\partial x_1} \\
\frac{\partial f}{\partial x_2} \\
\vdots \\
\frac{\partial f}{\partial x_n}
\end{bmatrix}.
$$

### Matrix Dimensions:
- $\nabla f \in \mathbb{R}^n$, matching $\mathbf{x}$'s shape.

---

## 3. **Comparison: Jacobian vs. Shape Convention**

### Jacobian:
- **Definition**: The Jacobian aligns with the mathematical convention of derivatives and facilitates composition using the chain rule.
- **Matrix Dimensions**: $\mathbf{J}_f \in \mathbb{R}^{1 \times n}$ for $f : \mathbb{R}^n \to \mathbb{R}$.
- **Application**: Ideal for computing derivatives in vectorized form; commonly used in symbolic calculus.

### Shape Convention:
- **Definition**: Adjusts the gradient's shape to match the input vector or parameter being updated.
- **Matrix Dimensions**: $\nabla f \in \mathbb{R}^n$, matching $\mathbf{x}$'s shape.
- **Application**: Essential in machine learning frameworks for gradient descent updates:
  $$
  \mathbf{x} \gets \mathbf{x} - \eta \nabla f,
  $$
  where $\eta$ is the learning rate.

---

## 4. **Practical Example: Outer Product in Gradients**

Consider a linear transformation $\mathbf{z} = \mathbf{W} \mathbf{x}$, where $\mathbf{W} \in \mathbb{R}^{m \times n}$, $\mathbf{x} \in \mathbb{R}^n$, and $\mathbf{z} \in \mathbb{R}^m$. Assume a scalar loss function $J$.

### Jacobian Perspective:
The Jacobian $\frac{\partial \mathbf{z}}{\partial \mathbf{W}}$ has dimensions $m \times (n \times m)$, forming a high-dimensional tensor, which is impractical for computation.

### Shape Convention:
The gradient with respect to $\mathbf{W}$ is reshaped directly:
$$
\frac{\partial J}{\partial \mathbf{W}} = \delta \mathbf{x}^\top,
$$
where $\delta \in \mathbb{R}^m$ is the backpropagated error.

### Matrix Dimensions:
- $\delta \mathbf{x}^\top \in \mathbb{R}^{m \times n}$, matching $\mathbf{W}$'s shape.
