## Key Idea
Key idea of RNN is that we have an internal state that is updated at each time step. This internal state is a function of the previous internal state and the current input. The internal state is then used to make a prediction.

## Mathematical Formulation of RNN

The mathematical formulation of RNN is as follows:
Assume that we have a sequence of inputs $x^{(1)}, x^{(2)}, \ldots, x^{(T)}$. Note each $x^{(t)}$ can be a vector and a scalar. For each associated input, we have a sequence of outputs $y^{(1)}, y^{(2)}, \ldots, y^{(T)}$. Again note that each $y^{(t)}$ can be a vector or a scalar. For example if we are predicting a stock price, $y^{(t)}$ can be a scalar. If we are predicting a sequence of words, $y^{(t)}$ can be a vector. Here we introduce hidden state $h^{(t)}$ which is a function of the previous hidden state and the current input. The hidden state is then used to make a prediction. At time 0, the hidden state $h_{0}$ is initialised. 

Now for each time step $t$, we look at the input at the current timestep denotated by $x_{t}$, we look at the previous hidden state $h_{t-1}$ and we compute the hidden state $h_{t}$ as follows:

$$h_{t} = f(Ux_{t} + Wh_{t-1}) + b$$

where $f$ is a non-linear activation function such as tanh or ReLU. Typically we use tanh in RNNs. $U$ and $W$ are the weight matrices. The output at time $t$ is computed as follows:

$$o_{t} = Vh_{t} + c$$

where $V$ is the weight matrix. In case we want to output a probability distribution prediction at time $t$ is computed as follows:

$$\hat{y}_{t} = softmax(o_{t})$$

where $\hat{y}_{t}$ is the predicted output at time $t$.
