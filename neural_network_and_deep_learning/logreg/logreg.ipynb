## **1. Binary Outcomes and Probabilities**
Logistic regression is used when the output $Y$ is binary, i.e., $Y \in \{0, 1\}$. The goal is to model the conditional probability $P(Y = 1 | \mathbf{X})$, where $\mathbf{X}$ is a vector of input features.

We define:
$$
p(\mathbf{X}) = P(Y = 1 | \mathbf{X})
$$

So $p(\mathbf{X})$ is the conditional probability of $Y = 1$ given $\mathbf{X}$.
Since probabilities are bounded between 0 and 1, direct modeling using a linear function $\mathbf{X} \cdot \boldsymbol{\beta}$ is unsuitable because it is unbounded.

---

## **2. The Logistic Function**
To ensure $p(\mathbf{X})$ lies within the range $[0, 1]$, the **logistic function**  is used:
$$
p(\mathbf{X}) = \frac{1}{1 + e^{-(\beta_0 + \mathbf{X} \cdot \boldsymbol{\beta})}}
$$

Note **Sigmoid Function**: is generally defined as $\frac{1}{1 + e^{-x}}$.
Hence the logistic function is a special case of the sigmoid function, where the linear combination $\beta_0 + \mathbf{X} \cdot \boldsymbol{\beta}$ is used as the input.
Hence 
$$
p(\mathbf{X}) = \frac{1}{1 + e^{-(\beta_0 + \mathbf{X} \cdot \boldsymbol{\beta})}}
= \sigma(\beta_0 + \mathbf{X} \cdot \boldsymbol{\beta})
$$




This formulation introduces two components:
- $\beta_0$: The intercept term.
- $\boldsymbol{\beta}$: The coefficient vector corresponding to the features in $\mathbf{X}$.

The logistic function transforms the linear combination $z = \beta_0 + \mathbf{X} \cdot \boldsymbol{\beta}$ into a probability.

---

## **3. Odds and Log-Odds (Logit) Transformation**
Instead of modeling probabilities directly, logistic regression models the **log-odds**:
$$
\text{logit}(p(\mathbf{X})) = \log \left( \frac{p(\mathbf{X})}{1 - p(\mathbf{X})} \right) = \beta_0 + \mathbf{X} \cdot \boldsymbol{\beta}
$$

Here:
-  $p(\mathbf{X})$ from $[0, 1]$ is the conditional probability of $Y = 1$ given $\mathbf{X}$.
- The logit function maps $p(\mathbf{X})$ from $[0, 1]$ to $(-\infty, \infty)$, enabling linear modeling.

---

## **4. Likelihood Function**
Logistic regression uses Maximum Likelihood Estimation (MLE) to estimate $\beta_0$ and $\boldsymbol{\beta}$. The likelihood for $n$ independent observations $\{(y_i, \mathbf{x}_i)\}$ is:
$$
L(\beta_0, \boldsymbol{\beta}) = \prod_{i=1}^n p(\mathbf{x}_i)^{y_i} (1 - p(\mathbf{x}_i))^{1 - y_i}
$$

Taking the natural logarithm (log-likelihood):
$$
\ell(\beta_0, \boldsymbol{\beta}) = \sum_{i=1}^n \left[ y_i \log(p(\mathbf{x}_i)) + (1 - y_i) \log(1 - p(\mathbf{x}_i)) \right]
$$



## **Logistic Regression MLE**

In logistic regression, the response variable $Y$ is binary $Y \in \{0, 1\}$. For a given set of features $\mathbf{x}_i$, the probability of $Y_i = 1$ is modeled using the logistic (sigmoid) function:

$$
p(\mathbf{x}_i) = \Pr(Y_i = 1 | \mathbf{x}_i) = \frac{1}{1 + e^{-(\beta_0 + \mathbf{x}_i^\top \boldsymbol{\beta})}}.
$$

For $Y_i = 0$, the probability is:

$$
\Pr(Y_i = 0 | \mathbf{x}_i) = 1 - p(\mathbf{x}_i).
$$

Thus, the conditional probability (or likelihood) for a single observation $(Y_i, \mathbf{x}_i)$ can be expressed compactly as:

$$
\Pr(Y_i | \mathbf{x}_i) = p(\mathbf{x}_i)^{Y_i} \cdot (1 - p(\mathbf{x}_i))^{1 - Y_i}.
$$

This formulation captures both cases ($Y_i = 1$ and $Y_i = 0$).

---

#### **Joint Probability of Observations**

Assuming the observations are independent, the joint probability (likelihood function) for the entire dataset $\{(Y_i, \mathbf{x}_i)\}_{i=1}^n$ is the product of the individual probabilities:

$$
L(\beta_0, \boldsymbol{\beta}) = \prod_{i=1}^n \Pr(Y_i | \mathbf{x}_i).
$$

Substituting the individual likelihood expression:

$$
L(\beta_0, \boldsymbol{\beta}) = \prod_{i=1}^n p(\mathbf{x}_i)^{Y_i} (1 - p(\mathbf{x}_i))^{1 - Y_i}.
$$

---

#### **Log-Likelihood Function**

To simplify optimization, the log of the likelihood function, known as the log-likelihood, is used:

$$
\ell(\beta_0, \boldsymbol{\beta}) = \log L(\beta_0, \boldsymbol{\beta}) = \sum_{i=1}^n \left[ Y_i \log(p(\mathbf{x}_i)) + (1 - Y_i) \log(1 - p(\mathbf{x}_i)) \right].
$$

Substituting $p(\mathbf{x}_i)$ into this expression:

$$
\ell(\beta_0, \boldsymbol{\beta}) = \sum_{i=1}^n \left[ Y_i (\beta_0 + \mathbf{x}_i^\top \boldsymbol{\beta}) - \log(1 + e^{\beta_0 + \mathbf{x}_i^\top \boldsymbol{\beta}}) \right].
$$

This form is more manageable for numerical optimization.


## Logistic Regression: Gradient Descent Update

---

## Step 1: Define the logistic (sigmoid) function

The logistic (or sigmoid) function is defined as

$$
\sigma(z) = \frac{1}{1 + e^{-z}}.
$$
In code or notation, we often write $h_\theta(x) = \sigma(\theta^T x)$.

---

## Step 2: Compute the derivative of the logistic function

We need the derivative of $\sigma(z)$ with respect to $z$. Recall that

$$
\sigma(z) = \frac{1}{1 + e^{-z}}, \quad
 1 - \sigma(z) = 1 - \frac{1}{1 + e^{-z}} = \frac{e^{-z}}{1 + e^{-z}}.
$$

Hence,

$$
\frac{d}{dz}\sigma(z) \;=\; \sigma(z)\,\bigl(1 - \sigma(z)\bigr).
$$
---

## Step 3: Define the cost function for logistic regression

In logistic regression, we use the following **cost function** (negative log-likelihood):

$$
J(\theta) = -\frac{1}{m}\sum_{i=1}^m \Bigl[y^{(i)} \ln\!\bigl(h_\theta(x^{(i)})\bigr) 
 \;+\; \bigl(1 - y^{(i)}\bigr)\,\ln\!\bigl(1 - h_\theta(x^{(i)})\bigr)\Bigr],
$$

where 
$$
 h_\theta(x^{(i)}) = \sigma\!\bigl(\theta^T x^{(i)}\bigr).
$$
Here, $m$ is the number of training examples, $x^{(i)}$ is the feature vector of the $i$th example, and $y^{(i)}$ is the corresponding label ($0$ or $1$).

---

## Step 4: Derive the gradient of the cost function

We want the gradient of $ J(\theta) $ with respect to each parameter $ \theta_j $. Let’s compute

$$
\frac{\partial}{\partial \theta_j} J(\theta).
$$

1. Start with the cost function term by term:

$$
   J(\theta) = -\frac{1}{m}\sum_{i=1}^m 
   \Bigl[y^{(i)} \ln\!\bigl(h_\theta(x^{(i)})\bigr) 
   \;+\; \bigl(1 - y^{(i)}\bigr)\,\ln\!\bigl(1 - h_\theta(x^{(i)})\bigr)\Bigr].
  $$

2. Notice that $h_\theta(x^{(i)}) = \sigma(\theta^T x^{(i)})$.  
   Let $z^{(i)} = \theta^T x^{(i)}$. Then $h_\theta(x^{(i)}) = \sigma(z^{(i)})$.

3. When we differentiate, we use the chain rule. For a single training example, consider the derivative of
  $$
   -\Bigl[y^{(i)}\ln\!\bigl(\sigma(z^{(i)})\bigr) 
   + (1 - y^{(i)})\ln\!\bigl(1 - \sigma(z^{(i)})\bigr)\Bigr].
   $$
   The result (after simplification) is:
   $$
   \sigma(z^{(i)}) - y^{(i)}.
   $$


4. Finally, we multiply by $x_j^{(i)}$ (the derivative of $z^{(i)} = \theta^T x^{(i)}$ with respect to $\theta_j$), then sum over all $i$, and include the factor $\frac{1}{m}$ from the cost function:

$$
   \frac{\partial}{\partial \theta_j} J(\theta)
   = \frac{1}{m} \sum_{i=1}^m \Bigl(\sigma(\theta^T x^{(i)}) - y^{(i)}\Bigr)\,x_j^{(i)}.
$$

Hence the **gradient** is:

$$
 \nabla J(\theta) 
 = \frac{1}{m} \sum_{i=1}^m \bigl(h_\theta(x^{(i)}) - y^{(i)})\,x^{(i)}.
$$
---

## Step 5: Gradient descent update

The gradient descent algorithm updates each parameter $\theta_j$ as follows:

$$
 \theta_j 
 := \theta_j 
 \;-\; \alpha \,\frac{\partial}{\partial \theta_j}J(\theta),
$$

 where $\alpha$ is the learning rate (a small positive constant).

Substituting our expression for $\frac{\partial}{\partial \theta_j}J(\theta)$, we get:

$$
 \theta_j 
 := \theta_j 
 \;-\; \alpha \;\frac{1}{m}\sum_{i=1}^m \Bigl(h_\theta(x^{(i)}) - y^{(i)}\Bigr)\,x_j^{(i)}.
$$

Equivalently, in vectorized form:

$$
 \theta 
 := \theta 
 \;-\; \alpha \;\frac{1}{m}\sum_{i=1}^m \Bigl(h_\theta(x^{(i)}) - y^{(i)}\Bigr)\,x^{(i)}.
$$
---

## Final update equation (summary)

- **Hypothesis**: $h_\theta(x) = \sigma(\theta^T x)$ with $\sigma(z) = \frac{1}{1 + e^{-z}}$.  
- **Gradient**:  
  $$
  \nabla J(\theta) = \frac{1}{m}\sum_{i=1}^m \Bigl(h_\theta(x^{(i)}) - y^{(i)})\,x^{(i)}.
  $$
- **Update rule** (for each $\theta_j$ or in vector form):  
  $$
  \theta := \theta \;-\; \alpha \nabla J(\theta).
  $$

That’s the derivation of the logistic regression gradient descent update equations!
