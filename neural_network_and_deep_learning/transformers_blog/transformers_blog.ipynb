## Introduction

In this blog, we will discuss the attention mechanism in transformers. 
We will illustrate the attention mechanism step by step using simple for-loop. 
Generally attention mechanism is illustrated using bunch of matrix multiplications. 
Different people have different learning styles. 
Some people understand the concept better with matrix multiplications as that is extremely concise, I believe looking at the
same concept from different angles helps in understanding the concept better.
NB: They way I will work through it is not the ideal way to implement attention mechanism in real world. The purpose of this blog is to understand the attention mechanism in transformers.


## Motivation

In the original transformer paper, Attention is all you need, the authors introduced the transformer model, as a way to solve for the translation problem. Transformer which evolved from RNN, LSTM, GRU, digests the input sequence and constructs a rich, contextualized embedding for each element. Attenion mechanism in transformer model can be seen in general as a way to capture detailed summaries of not just the word itself but also its relationships with every other word in the sequence. Next work prediction, sentiment analysis, machine translation, image captioning, etc are some of the tasks which can be achieved as a by product of transformer model which contains all the structural and contextual information needed. 


## Lets Begin

Lets start with an input sequence. 

::: {.callout-note appearance="simple"}
Jack and Jill went up the hill \<eos\>.
:::

As a first step we need to tokenise the sentence. For simplicity we will use words as tokens. There are 8 tokens in the sentence, also notice that we have added end of sentence token <eos> towards the end of the sentence. There are bunch of other special tokens that are actually used that we can discuss in future. Note a encoder in transformer model maps $(x_1, x_2, \ldots x_n)$ to a sequence of continuous representations $(z_1, z_2, \ldots z_n)$.
